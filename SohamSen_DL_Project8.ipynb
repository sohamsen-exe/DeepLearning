#Step 1: Import libraries and create a dataset

```python
import numpy as np
import matplotlib.pyplot as plt

# Simple dataset
X = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([3, 5, 7, 9, 11], dtype=float)

# Number of data points
n = len(X)

# Display dataset
print("X:", X)
print("y:", y)
```

#Step 2: Initialize model parameters

```python
# Initialize parameters
w = 0.0       # weight
b = 0.0       # bias

# Learning rate
learning_rate = 0.01

# Number of iterations
iterations = 1000

# Store loss values
loss_history = []
```

#Step 3: Implement Gradient Descent

```python
for i in range(iterations):

    # Make predictions
    y_pred = w * X + b

    # Calculate error
    error = y_pred - y

    # Calculate Mean Squared Error
    loss = np.mean(error ** 2)

    # Calculate gradients
    dw = (2 / n) * np.sum(X * error)
    db = (2 / n) * np.sum(error)

    # Update parameters
    w = w - learning_rate * dw
    b = b - learning_rate * db

    # Store loss
    loss_history.append(loss)

    # Display progress every 100 iterations
    if (i + 1) % 100 == 0:
        print("Iteration:", i + 1,
              "Weight:", round(w, 4),
              "Bias:", round(b, 4),
              "Loss:", round(loss, 4))
```

#Step 4: Display the final parameters

```python
print("\nFinal Parameters:")
print("Weight (w):", w)
print("Bias (b):", b)

print("\nFinal Predictions:")
print(y_pred)
```

#Step 5: Visualize the decrease in loss

```python
plt.figure(figsize=(8, 5))

plt.plot(loss_history)

plt.xlabel("Iteration")
plt.ylabel("Loss")
plt.title("Loss Reduction During Gradient Descent")

plt.grid(True)
plt.show()
```

#Step 6: Visualize the fitted line

```python
plt.figure(figsize=(8, 5))

plt.scatter(X, y, label="Actual Data")
plt.plot(X, w * X + b, label="Predicted Line")

plt.xlabel("X")
plt.ylabel("y")
plt.title("Linear Regression Using Gradient Descent")

plt.legend()
plt.grid(True)
plt.show()
```

#Step 7: Show parameter updates

```python
# Reset parameters
w = 0.0
b = 0.0

learning_rate = 0.01

print("Parameter updates:\n")

for i in range(10):

    y_pred = w * X + b
    error = y_pred - y

    loss = np.mean(error ** 2)

    dw = (2 / n) * np.sum(X * error)
    db = (2 / n) * np.sum(error)

    w = w - learning_rate * dw
    b = b - learning_rate * db

    print("Iteration:", i + 1)
    print("Weight:", round(w, 4))
    print("Bias:", round(b, 4))
    print("Loss:", round(loss, 4))
    print()
```

Gradient Descent successfully optimized the weight and bias parameters of the linear regression model. As the number of iterations increased, the loss decreased, indicating that the model's predictions were becoming closer to the actual values. This demonstrates how iterative parameter updates can minimize prediction error during model training.
