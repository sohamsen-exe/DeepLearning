Students will calculate different loss values and understand the role of loss functions in measuring model prediction errors.



```python
import numpy as np
```

```python
# Actual class labels
y_actual = np.array([1, 0, 1, 1, 0])

# Predicted probabilities
y_predicted = np.array([0.9, 0.2, 0.7, 0.8, 0.1])

print("Actual Labels:", y_actual)
print("Predicted Values:", y_predicted)
```

```python
# Calculate Mean Squared Error
mse = np.mean((y_actual - y_predicted) ** 2)

print("Mean Squared Error:", mse)
```

```python
# Prevent log(0) errors
epsilon = 1e-15

y_predicted = np.clip(y_predicted, epsilon, 1 - epsilon)

# Calculate Cross-Entropy Loss
cross_entropy = -np.mean(
    y_actual * np.log(y_predicted) +
    (1 - y_actual) * np.log(1 - y_predicted)
)

print("Cross-Entropy Loss:", cross_entropy)
```

```python
print("Loss Function Comparison\n")
print("Mean Squared Error:", mse)
print("Cross-Entropy Loss:", cross_entropy)
```

MSE calculates the squared difference between actual and predicted values.
Cross-Entropy measures the difference between actual classes and predicted probabilities.

MSE is commonly used for regression problems.
Cross-Entropy is commonly used for classification problems.
