# Project 5: Activation Functions Using Python

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def ReLu(x):
    return np.maximum(0, x)

def tanh(x):
    return np.tanh(x)
```

```python
inputs = np.array([-5, -3, -2, -1, 0, 1, 2, 3, 5])
```

```python
sigmoid_output = sigmoid(inputs)
relu_output = ReLu(inputs)
tanh_output = tanh(inputs)
```

```python
print("Input\tSigmoid\t\tReLU\t\tTanh")

for x, s, r, t in zip(inputs, sigmoid_output, relu_output, tanh_output):
    print(f"{x}\t{s:.4f}\t\t{r:.4f}\t\t{t:.4f}")
```

```python
x = np.linspace(-5, 5, 200)
```

```python
y_sigmoid = sigmoid(x)
y_relu = ReLu(x)
y_tanh = tanh(x)
```

```python
plt.figure(figsize=(12, 8))

# Sigmoid
plt.subplot(3, 1, 1)
plt.plot(x, y_sigmoid, linewidth=2)
plt.title("Sigmoid Activation Function")
plt.xlabel("Input (x)")
plt.ylabel("Sigmoid(x)")
plt.grid(True)

# ReLU
plt.subplot(3, 1, 2)
plt.plot(x, y_relu, linewidth=2)
plt.title("ReLU Activation Function")
plt.xlabel("Input (x)")
plt.ylabel("ReLU(x)")
plt.grid(True)

# Tanh
plt.subplot(3, 1, 3)
plt.plot(x, y_tanh, linewidth=2)
plt.title("Tanh Activation Function")
plt.xlabel("Input (x)")
plt.ylabel("Tanh(x)")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.plot(x, y_sigmoid, label="Sigmoid", linewidth=2)
plt.plot(x, y_relu, label="ReLU", linewidth=2)
plt.plot(x, y_tanh, label="Tanh", linewidth=2)

plt.title("Comparison of Activation Functions")
plt.xlabel("Input (x)")
plt.ylabel("Output")
plt.axhline(0, linewidth=1)
plt.axvline(0, linewidth=1)
plt.legend()
plt.grid(True)

plt.show()
```

```python
import pandas as pd

data = {
    "Input": inputs,
    "Sigmoid": sigmoid_output,
    "ReLU": relu_output,
    "Tanh": tanh_output
}

df = pd.DataFrame(data)

df = df.round(4)

display(df)
```
